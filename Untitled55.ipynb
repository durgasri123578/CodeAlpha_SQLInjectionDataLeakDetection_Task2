# Detecting Data Leaks Using SQL Injection

## Task 2

This project protects user data from SQL injection.

It uses a database to store user information.

## Step 1: Create Database

Here, we create a simple database for our project.

```python
import sqlite3

# Create database
connection = sqlite3.connect("secure_users.db")

# Create cursor
cursor = connection.cursor()

print("Database created successfully.")
```

## Step 2: Create User Table

Here, we create a table to store user details.

The table contains a user ID, username, and password.

```python
# Create users table

cursor.execute("""
CREATE TABLE IF NOT EXISTS users (
    user_id INTEGER PRIMARY KEY,
    username TEXT UNIQUE,
    password TEXT
)
""")

connection.commit()

print("User table created successfully.")
```

## Step 3: Add Sample User

Here, we add a sample user to the database.

This user will be used to test the login system.

```python
# Add sample user

cursor.execute("""
INSERT OR IGNORE INTO users (username, password)
VALUES (?, ?)
""", ("Arun", "Arun@123"))

connection.commit()

print("Sample user added successfully.")
```

## Step 4: Create Secure Login

Here, we check the username and password.

We use a safe SQL query to protect the database.

```python
# Secure login function

def secure_login(username, password):

    query = """
    SELECT * FROM users
    WHERE username = ? AND password = ?
    """

    cursor.execute(query, (username, password))

    user = cursor.fetchone()

    if user:
        print("Login successful.")
    else:
        print("Invalid username or password.")


secure_login("Arun", "Arun@123")
```

## Step 5: Validate User Input

Here, we check the username and password before sending them to the database.

Empty input will not be accepted.

```python
def validate_input(username, password):

    if username.strip() == "" or password.strip() == "":
        print("Username and password cannot be empty.")
        return False

    return True


username = "Arun"
password = "Arun@123"

if validate_input(username, password):
    secure_login(username, password)
```

## Step 6: Add Access Code

Here, we add an access code for extra security.

Only users with the correct access code can use the system.

```python
# Access code

ACCESS_CODE = "CLOUD@123"

user_code = "CLOUD@123"

if user_code == ACCESS_CODE:
    print("Access granted.")
else:
    print("Access denied.")
```

## Step 7: Encrypt Sensitive Data

Here, we encrypt sensitive data using AES encryption.

The encrypted data cannot be read directly.

```python
from cryptography.fernet import Fernet

# Create an encryption key
key = Fernet.generate_key()

cipher = Fernet(key)

# Data to encrypt
sensitive_data = "Arun@123"

# Encrypt the data
encrypted_data = cipher.encrypt(sensitive_data.encode())

print("Encrypted data:", encrypted_data)
```

## Step 8: Decrypt Data

Here, we decrypt the encrypted data.

The original data can be recovered using the correct key.

```python
# Decrypt the data

decrypted_data = cipher.decrypt(encrypted_data).decode()

print("Decrypted data:", decrypted_data)
```

## Step 9: Test Invalid Login

Here, we test the login with a wrong password.

The system should reject the login.

```python
# Test invalid login

username = "Arun"
password = "Wrong@123"

if validate_input(username, password):
    secure_login(username, password)
```

## Step 10: Test Access Code

Here, we test the access code.

The system allows access only when the code is correct.

```python
# Test access code

user_code = "WRONG@123"

if user_code == ACCESS_CODE:
    print("Access granted.")
else:
    print("Access denied.")
```

## Step 11: Test Safe SQL Query

Here, we use a parameterized SQL query.

This helps protect the database from SQL injection.

```python
# Test safe SQL query

username = "Arun"

query = "SELECT * FROM users WHERE username = ?"

cursor.execute(query, (username,))

result = cursor.fetchone()

if result:
    print("User found.")
else:
    print("User not found.")
```

## Step 12: Close the Database

Here, we close the database connection after completing the work.

```python
# Close the database connection

connection.close()

print("Database connection closed successfully.")
```
